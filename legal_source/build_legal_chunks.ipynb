```python
import pandas as pd

# load legal staging data
df = pd.read_csv("legal_clauses_staging.csv")

rows = []

for i, row in df.iterrows():
    text = str(row["clause_text"]).strip()
    token_estimate = len(text.split())  # simple, consistent with your current data

    rows.append({
        "chunk_id": f"LC{i+1:05d}",
        "doc_id": f"LEGAL_DOC_{(i//10)+1:03d}",
        "domain": "legal",
        "chunk_index": 0,
        "estimated_tokens": token_estimate,
        "chunk_text": text
    })

legal_chunks_df = pd.DataFrame(rows)

# save to legal data folder
legal_chunks_df.to_csv("../data_legal/rag_corpus_chunks.csv", index=False)

print(f"Created {len(legal_chunks_df)} legal chunks")
```
