```python
import pandas as pd

# paths
INPUT_FILE = "dataset_b_covenants.csv"
OUTPUT_FILE = "legal_clauses_staging.csv"

# load data
df = pd.read_csv(INPUT_FILE)

rows = []

def clean_text(x):
    if pd.isna(x):
        return None
    if isinstance(x, str):
        x = x.strip()
        if x == "[]" or x == "":
            return None
        return x.strip("[]").strip("'")
    return None

CLAUSE_MAP = {
    "Most Favored Nation": "most_favored_nation",
    "Non-Compete": "non_compete",
    "Exclusivity": "exclusivity",
    "No-Solicit Of Customers": "no_solicit_customers",
    "No-Solicit Of Employees": "no_solicit_employees",
    "Non-Disparagement": "non_disparagement",
}

for _, row in df.iterrows():
    document_name = row["Document Name"]

    for col, clause_type in CLAUSE_MAP.items():
        answer_col = f"{col}-Answer"
        if row.get(answer_col) == "Yes":
            clause_text = clean_text(row.get(col))
            if clause_text:
                rows.append({
                    "source_file": "dataset_b_covenants.csv",
                    "document_name": document_name,
                    "clause_type": clause_type,
                    "clause_text": clause_text
                })

out_df = pd.DataFrame(rows)
out_df.to_csv(OUTPUT_FILE, index=False)

print(f"Extracted {len(out_df)} clauses")
```
