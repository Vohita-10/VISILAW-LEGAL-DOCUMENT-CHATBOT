```python
import pandas as pd
chunks = pd.read_csv("../data_legal/rag_corpus_chunks.csv")
```

```python
# Create stable row_id(This is the only thing BM25 and FAISS must agree on.)
# both indexes return row_ids, they are then rank normalized, merged, and top-k are returned. Based on row_id chunk_id is fetched.
chunks = chunks.reset_index(drop=True)
chunks["row_id"] = chunks.index
```

```python
from src.retrieval.tokenization import tokenize

tokenized_corpus = [
    tokenize(text)
    for text in chunks["chunk_text"]
]
```

```python
from pathlib import Path
import pickle
from src.retrieval.bm25 import BM25Index

# build BM25 index (LEGAL)
bm25_index = BM25Index(
    documents=chunks["chunk_text"].astype(str).tolist()
)

# LEGAL index directory
INDEX_DIR = Path.cwd().parent / "data_legal" / "indexes"
INDEX_DIR.mkdir(parents=True, exist_ok=True)

# save LEGAL BM25 index
index_path = INDEX_DIR / "bm25_legal.pkl"
with open(index_path, "wb") as f:
    pickle.dump(bm25_index, f)

index_path

```
